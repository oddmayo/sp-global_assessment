# Sources

Financial data like this is not by day to day. I had to read quite a lot to start tackling the problem even though I had experience with time series.

This is are the main ones that helped me:


#### **Financial Machine Learning & Time-Series Methodology**

- **Financial ML Principles**: De Prado, M. L. (2018). *Advances in Financial Machine Learning*. John Wiley & Sons.  

    Contains valuable information for doing chronological splits and stationarity transformations

- **Empirical Asset Pricing**: Gu, S., Kelly, B., & Xiu, D. (2020). *Empirical Asset Pricing via Machine Learning*. The Review of Financial Studies, 33(5), 2223-2273.

    Explains why regularized linear models outperform complex non-linear ones on financial price levels.

#### **Technical Indicators & Feature Engineering**

- **Technical Analysis**: Murphy, J. J. (1999). *Technical Analysis of the Financial Markets*. New York Institute of Finance.  
    Guide for constructing standard momentum (RSI-14), trend (SMAs), and volatility features.

#### Financial NLP & Sentiment Analysis

- **FinBERT**: Araci, D. (2019). *FinBERT: Financial Sentiment Analysis with Pre-trained Language Models*. arXiv preprint arXiv:1908.10063.  
    Primary reference for using domain-adapted BERT models to extract sentiment from financial news and text.
- **VADER Sentiment**: Hutto, C. J., & Gilbert, E. E. (2014). *VADER: A Parsimonious Rule-based Model for Sentiment Analysis of Social Media Text*. Proceedings of the International AAAI Conference on Web and Social Media, 8(1), 216-225.  
  Reference for lightweight, rule-based sentiment scoring.
